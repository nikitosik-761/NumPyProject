# Круговая диаграмма


1. Постройте круговую диаграмму в Python, используя Matplotlib
2. Как установить границу для клиньев в круговой диаграмме Matplotlib?
3. Радиально сместите клин круговой диаграммы в Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt

categories = np.array(["Juniors", "Middles", "Seniors", "Teamleads"])
counts = np.array([20, 13, 3, 1])
plt.pie(counts,
        labels=categories,
        explode=[0, 0, 0, 0.2],  # 3 смещение клинаб
        wedgeprops={ # установка границ для клиньев
            "edgecolor": "black",
            "linewidth": 0.5,
            "linestyle": "-"
        })
plt.show()
```
